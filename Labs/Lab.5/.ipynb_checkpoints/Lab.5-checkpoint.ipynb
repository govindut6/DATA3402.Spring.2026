1. Create a `matrix` class with the following properties:
    * It can be initialized in 2 ways:
        1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    * Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
    * Matrix assignment works in 2 ways:
        1. If `M_1` and `M_2` are `matrix` instances `M_1=M_2` sets the values of `M_1` to those of `M_2`, if they are the same size. Error otherwise.
        2. In example above `M_2` can be a list of lists of correct size.


```python
# Solution
class Matrix:
    def __init__(self, *args):
        if len(args) == 2:  # Matrix(n, m)
            n, m = args
            self.data = [[0.0]*m for _ in range(n)]

        elif len(args) == 1 and isinstance(args[0], list):  # Matrix(list_of_lists)
            rows = args[0]
            if not all(len(row) == len(rows[0]) for row in rows):
                raise ValueError("Rows must have same length")
            self.data = [[float(x) for x in row] for row in rows]

        else:
            raise ValueError("Invalid input")

    def shape(self):
        return len(self.data), len(self.data[0])

    def __getitem__(self, key):
        if isinstance(key, tuple):
            i, j = key
            return self.data[i][j]
        return self.data[key]

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.data[i][j] = float(value)
        else:
            self.data[key] = value

    def assign(self, other):
        if isinstance(other, Matrix):
            if self.shape() != other.shape():
                raise ValueError("Size mismatch")
            self.data = [row[:] for row in other.data]
        elif isinstance(other, list):
            if (len(other), len(other[0])) != self.shape():
                raise ValueError("Size mismatch")
            self.data = [[float(x) for x in row] for row in other]
        else:
            raise ValueError("Invalid assignment")

    def __repr__(self):
        return str(self.data)
```

```python
# Test
M1 = Matrix(2,3)
M2 = Matrix([[1,2,3],[4,5,6]])

print(M2[1,2])
M1.assign(M2)
print(M1)

try:
    Matrix([[1,2],[3]])
except:
    print("Invalid matrix caught")
```

2. Add the following methods:
    * `shape()`: returns a tuple `(n,m)` of the shape of the matrix.
    * `transpose()`: returns a new matrix instance which is the transpose of the matrix.
    * `row(n)` and `column(n)`: that return the nth row or column of the matrix M as a new appropriately shaped matrix object.
    * `to_list()`: which returns the matrix as a list of lists.
    *  `block(n_0,n_1,m_0,m_1)` that returns a smaller matrix located at the n_0 to n_1 columns and m_0 to m_1 rows. 
    * (Extra credit) Modify `__getitem__` implemented above to support slicing.
        

```python
# Solution
class Matrix:
    def __init__(self, *args):
        if len(args) == 2:
            n, m = args
            self.data = [[0.0]*m for _ in range(n)]

        elif len(args) == 1 and isinstance(args[0], list):
            rows = args[0]
            if not all(len(row) == len(rows[0]) for row in rows):
                raise ValueError("Rows must have same length")
            self.data = [[float(x) for x in row] for row in rows]
        else:
            raise ValueError("Invalid input")

    def shape(self):
        return len(self.data), len(self.data[0])

    def transpose(self):
        return Matrix([[self.data[i][j] for i in range(len(self.data))]
                       for j in range(len(self.data[0]))])

    def row(self, n):
        return Matrix([self.data[n]])

    def column(self, n):
        return Matrix([[row[n]] for row in self.data])

    def to_list(self):
        return [row[:] for row in self.data]

    def block(self, n0, n1, m0, m1):
        return Matrix([row[n0:n1] for row in self.data[m0:m1]])

    def __getitem__(self, key):
        if isinstance(key, tuple):
            r, c = key
            if isinstance(r, slice) or isinstance(c, slice):
                rows = self.data[r]
                return Matrix([row[c] for row in rows])
            return self.data[r][c]
        return self.data[key]

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.data[i][j] = float(value)
        else:
            self.data[key] = value

    def assign(self, other):
        if isinstance(other, Matrix):
            if self.shape() != other.shape():
                raise ValueError("Size mismatch")
            self.data = [row[:] for row in other.data]
        elif isinstance(other, list):
            if (len(other), len(other[0])) != self.shape():
                raise ValueError("Size mismatch")
            self.data = [[float(x) for x in row] for row in other]
        else:
            raise ValueError("Invalid assignment")

    def __repr__(self):
        return str(self.data)
```

```python
# Test
M = Matrix([[1,2,3],
            [4,5,6],
            [7,8,9]])

print("Shape:", M.shape())

print("Transpose:", M.transpose())

print("Row 1:", M.row(1))

print("Column 2:", M.column(2))

print("To list:", M.to_list())

print("Block (cols 0-2, rows 0-2):", M.block(0,2,0,2))

# Extra credit slicing
print("Slice:", M[0:2, 1:3])
```

3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python
# Solution
def constant(n, m, c):
    return Matrix([[float(c)] * m for _ in range(n)])

def zeros(n, m):
    return constant(n, m, 0.0)

def ones(n, m):
    return constant(n, m, 1.0)

def eye(n):
    return Matrix([[1.0 if i == j else 0.0 for j in range(n)] for i in range(n)])
```

```python
# Test
print("Constant:")
print(constant(2,3,5))

print("Zeros:")
print(zeros(2,3))

print("Ones:")
print(ones(2,3))

print("Identity:")
print(eye(3))
```

4. Add the following member functions to your class. Make sure to appropriately test the dimensions of the matrices to make sure the operations are correct.
    * `M.scalarmul(c)`: a matrix that is scalar product $cM$, where every element of $M$ is multiplied by $c$.
    * `M.add(N)`: adds two matrices $M$ and $N$. Don’t forget to test that the sizes of the matrices are compatible for this and all other operations.
    * `M.sub(N)`: subtracts two matrices $M$ and $N$.
    * `M.mat_mult(N)`: returns a matrix that is the matrix product of two matrices $M$ and $N$.
    * `M.element_mult(N)`: returns a matrix that is the element-wise product of two matrices $M$ and $N$.
    * `M.equals(N)`: returns true/false if $M==N$.

```python
# Solution
class Matrix:
    def __init__(self, *args):
        if len(args) == 2:
            n, m = args
            self.data = [[0.0]*m for _ in range(n)]
        elif len(args) == 1 and isinstance(args[0], list):
            rows = args[0]
            if not all(len(row) == len(rows[0]) for row in rows):
                raise ValueError("Rows must have same length")
            self.data = [[float(x) for x in row] for row in rows]
        else:
            raise ValueError("Invalid input")

    def shape(self):
        return len(self.data), len(self.data[0])

    def __getitem__(self, key):
        if isinstance(key, tuple):
            r, c = key
            return self.data[r][c]
        return self.data[key]

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.data[i][j] = float(value)
        else:
            self.data[key] = value

    def assign(self, other):
        if self.shape() != other.shape():
            raise ValueError("Size mismatch")
        self.data = [row[:] for row in other.data]

    # ===== Q4 METHODS =====

    def scalarmul(self, c):
        return Matrix([[c * val for val in row] for row in self.data])

    def add(self, N):
        if self.shape() != N.shape():
            raise ValueError("Size mismatch")
        return Matrix([[self.data[i][j] + N.data[i][j]
                        for j in range(self.shape()[1])]
                        for i in range(self.shape()[0])])

    def sub(self, N):
        if self.shape() != N.shape():
            raise ValueError("Size mismatch")
        return Matrix([[self.data[i][j] - N.data[i][j]
                        for j in range(self.shape()[1])]
                        for i in range(self.shape()[0])])

    def element_mult(self, N):
        if self.shape() != N.shape():
            raise ValueError("Size mismatch")
        return Matrix([[self.data[i][j] * N.data[i][j]
                        for j in range(self.shape()[1])]
                        for i in range(self.shape()[0])])

    def mat_mult(self, N):
        if self.shape()[1] != N.shape()[0]:
            raise ValueError("Incompatible dimensions")
        return Matrix([[sum(self.data[i][k] * N.data[k][j]
                        for k in range(self.shape()[1]))
                        for j in range(N.shape()[1])]
                        for i in range(self.shape()[0])])

    def equals(self, N):
        return self.data == N.data

    def __repr__(self):
        return str(self.data)
```

```python
# Test
A = Matrix([[1,2],[3,4]])
B = Matrix([[5,6],[7,8]])

print("Scalar:", A.scalarmul(2))
print("Add:", A.add(B))
print("Subtract:", A.sub(B))
print("Element-wise:", A.element_mult(B))
print("Matrix multiply:", A.mat_mult(B))
print("Equals:", A.equals(B))
```

5. Overload python operators to appropriately use your functions in 4 and allow expressions like:
    * 2*M
    * M*2
    * M+N
    * M-N
    * M*N
    * M==N
    * M=N


```python
# Solution
class Matrix:
    def __init__(self, *args):
        if len(args) == 2:
            n, m = args
            self.data = [[0.0]*m for _ in range(n)]
        elif len(args) == 1 and isinstance(args[0], list):
            rows = args[0]
            if not all(len(row) == len(rows[0]) for row in rows):
                raise ValueError("Rows must have same length")
            self.data = [[float(x) for x in row] for row in rows]
        else:
            raise ValueError("Invalid input")

    def shape(self):
        return len(self.data), len(self.data[0])

    def __getitem__(self, key):
        if isinstance(key, tuple):
            r, c = key
            return self.data[r][c]
        return self.data[key]

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.data[i][j] = float(value)
        else:
            self.data[key] = value

    def scalarmul(self, c):
        return Matrix([[c * val for val in row] for row in self.data])

    def add(self, N):
        if self.shape() != N.shape():
            raise ValueError("Size mismatch")
        return Matrix([[self.data[i][j] + N.data[i][j]
                        for j in range(self.shape()[1])]
                        for i in range(self.shape()[0])])

    def sub(self, N):
        if self.shape() != N.shape():
            raise ValueError("Size mismatch")
        return Matrix([[self.data[i][j] - N.data[i][j]
                        for j in range(self.shape()[1])]
                        for i in range(self.shape()[0])])

    def element_mult(self, N):
        if self.shape() != N.shape():
            raise ValueError("Size mismatch")
        return Matrix([[self.data[i][j] * N.data[i][j]
                        for j in range(self.shape()[1])]
                        for i in range(self.shape()[0])])

    def mat_mult(self, N):
        if self.shape()[1] != N.shape()[0]:
            raise ValueError("Incompatible dimensions")
        return Matrix([[sum(self.data[i][k] * N.data[k][j]
                        for k in range(self.shape()[1]))
                        for j in range(N.shape()[1])]
                        for i in range(self.shape()[0])])

    def equals(self, N):
        return self.data == N.data

    # ===== OPERATOR OVERLOADING =====
    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return self.scalarmul(other)
        if isinstance(other, Matrix):
            return self.mat_mult(other)
        raise ValueError("Invalid multiplication")

    def __rmul__(self, other):
        return self.scalarmul(other)

    def __add__(self, other):
        return self.add(other)

    def __sub__(self, other):
        return self.sub(other)

    def __eq__(self, other):
        return self.equals(other)

    def __repr__(self):
        return str(self.data)
```

6. Demonstrate the basic properties of matrices with your matrix class by creating two 2 by 2 example matrices using your Matrix class and illustrating the following:

$$
(AB)C=A(BC)
$$
$$
A(B+C)=AB+AC
$$
$$
AB\neq BA
$$
$$
AI=A
$$

```python
# Create 2x2 matrices
A = Matrix([[1,2],
            [3,4]])

B = Matrix([[0,1],
            [1,0]])

C = Matrix([[2,0],
            [0,2]])

I = eye(2)  # Identity matrix from Question 3


# 1 question
left1 = (A * B) * C
right1 = A * (B * C)

print("(AB)C:", left1)
print("A(BC):", right1)
print("Associative Property Holds:", left1 == right1)


# 2 question
left2 = A * (B + C)
right2 = (A * B) + (A * C)

print("\nA(B+C):", left2)
print("AB + AC:", right2)
print("Distributive Property Holds:", left2 == right2)


# 3 question
print("\nAB:", A * B)
print("BA:", B * A)
print("AB != BA:", (A * B) != (B * A))


# 4 question
print("\nAI:", A * I)
print("A:", A)
print("Identity Property Holds:", (A * I) == A)
```
